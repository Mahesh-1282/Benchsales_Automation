```python
%sql
-- 1. Create Database/Schema
CREATE CATALOG IF NOT EXISTS jobs_automation_db;
```

```python
%sql
USE jobs_automation_db.default;
```

```python
%sql
-- 2. Users Master Table
CREATE TABLE IF NOT EXISTS users_master (
    user_id STRING,
    full_name STRING,
    email_address STRING,
    phone_number STRING,
    linkedin_url STRING,
    github_url STRING,
    created_at TIMESTAMP,
    is_active BOOLEAN
) USING DELTA;
```

```python
%sql
-- 3. Parsed Profiles Table
CREATE TABLE IF NOT EXISTS user_parsed_profiles (
    profile_id STRING,
    user_id STRING,
    target_role STRING,
    original_resume_text STRING,
    parsed_skills_json STRING,
    parsed_experience_json STRING,
    parsed_projects_json STRING
) USING DELTA;
```

```python
%sql
-- 4. Raw Scraped Jobs Table
CREATE TABLE jobs_automation_db.default.raw_jobs_staging (
    id STRING,                  -- Unique UUID for each row
    job_hash STRING,            -- MD5 URL Hash to prevent duplicates
    created_at TIMESTAMP,       -- Time when job was added
    updated_at TIMESTAMP,       -- Time when job was modified
    fetch_date DATE,            -- For easy daily filtering
    search_keyword STRING,      -- E.g., Data Engineer
    company_name STRING,        -- Cleaned Company Name
    job_title STRING,           -- Cleaned EXACT Job Title
    job_description STRING,     -- Deep Scraped Pure Text
    apply_link STRING,          -- Direct Apply URL
    validation_status STRING    -- Pending, Processed
) USING DELTA;
```

```python
%sql
-- 5. Validated Golden Jobs Table
CREATE TABLE jobs_automation_db.default.validated_jobs_master (
    job_id STRING,
    company_name STRING,
    job_title STRING,
    job_type STRING,            -- W2, C2C, Full-Time, Contract, or Not Found
    is_valid BOOLEAN,
    hr_email STRING,
    client_requirements STRING, -- AI రాసిన ఎగ్జాక్ట్ రిక్వైర్మెంట్స్
    apply_url STRING,
    validated_date TIMESTAMP
) USING DELTA;
```

```python
%sql
-- 6. Application & Interview Tracking Table
CREATE TABLE IF NOT EXISTS application_tracking_analytics (
    application_id STRING,
    user_id STRING,
    job_id STRING,
    generated_resume_path STRING,
    action_taken STRING,
    applied_date TIMESTAMP,
    interview_status STRING,
    last_email_synced_at TIMESTAMP
) USING DELTA;
```
