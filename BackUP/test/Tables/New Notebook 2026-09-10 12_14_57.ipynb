```python
# File path ni mee path tho replace cheyyandi
file_path = "/Workspace/Users/maheshmahi1282@gmail.com/Benchsales_Automation_Upgraded_Version/test/jobs_v10_output_test_2.csv"

# Read the CSV file into a PySpark DataFrame
jobs_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("multiLine", "true") \
    .option("escape", '"') \
    .load(file_path)

# Data correct ga vachindo ledo check cheyyadaniki
display(jobs_df)
```

```python
table_name = "jobs_automation_db.default.jobs_harvested_bronze"

# Write the DataFrame to a Delta Lake table
jobs_df.write.format("delta") \
    .mode("append") \
    .option("mergeSchema", "true") \
    .saveAsTable(table_name)

print(f"Data successfully loaded into {table_name}")
```
